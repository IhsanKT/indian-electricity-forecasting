## Full period

The visible upward drift is real demand growth, not an artefact.

```python
eda.fig_full_period(y)
```

## Daily shape

Trough around 03:00, afternoon peak, second evening peak.

```python
eda.fig_hour_of_day(y)
```

## Day of week

Sunday sits clearly below the working week — this is the effect the **weekly** seasonal naive exploits and the daily one misses.

```python
eda.fig_day_of_week(y)
```

## Weekly seasonality

```python
eda.fig_weekly_profile(y)
```

## Annual seasonality

Summer peak (May–June), monsoon dip (July–August), November trough. Monsoon-driven seasonality is one of the features this project hypothesises Western-pretrained models may under-represent.

```python
eda.fig_monthly(y)
```

## Distribution

```python
eda.fig_distribution(y)
```

## Missing data

Gaps before forward-filling. Fills use past information only — interpolating across a gap with later values would be leakage.

```python
eda.fig_missing_map(y)
```
